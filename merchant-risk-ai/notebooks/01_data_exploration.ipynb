# Synthetic transaction data exploration

This notebook inspects the Stage 2 synthetic dataset only. It does not split data, select a model, calculate a decision threshold, or report model performance.

```python
from pathlib import Path

import pandas as pd

dataset_path = Path("../data/raw/synthetic_transactions.csv")
transactions = pd.read_csv(dataset_path)
transactions.shape, transactions.head()
```

```python
# Schema, missingness, and label balance
transactions.info()
display(transactions.isna().sum().rename("missing_values"))
display(transactions["is_high_risk"].value_counts().rename("count"))
transactions["is_high_risk"].value_counts(normalize=True).rename("share")
```

```python
# Descriptive comparisons only; these are not model metrics.
display(transactions.groupby("is_high_risk")[["order_amount", "account_age_days", "failed_payment_attempts_24h"]].agg(["count", "mean", "median"]))
display(pd.crosstab(transactions["is_high_risk"], transactions["payment_method"], normalize="index"))
```
